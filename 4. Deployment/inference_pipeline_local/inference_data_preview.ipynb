# Inference Data Preview

Loads data from Redshift into a pandas DataFrame for quick inspection.

```python
import os
import pandas as pd
import yaml
from dotenv import load_dotenv
import psycopg2

CFG_PATH = "../training_pipeline_simplified/config.yaml"
ENV_PATH = "../training_pipeline_simplified/.env"

with open(CFG_PATH, "r") as f:
    cfg = yaml.safe_load(f)

rs = cfg["data"]["redshift_kwargs"]
load_dotenv(ENV_PATH)
password = rs["password"]
if password.startswith("${") and password.endswith("}"):
    password = os.getenv(password[2:-1])

if not password:
    raise RuntimeError("REDSHIFT_PASSWORD not found in .env")

conn = psycopg2.connect(
    host=rs["host"],
    port=rs["port"],
    dbname=rs["dbname"],
    user=rs["user"],
    password=password,
)
```

```python
# Choose date column for "today only" filter
DATE_COL = "last_contact_ts"  # timestamp column in inference_features

sql = f"""
SELECT *
FROM dth_churn_ml_inference.inference_features
LIMIT 200;
"""

df = pd.read_sql(sql, conn)
df.head()
```

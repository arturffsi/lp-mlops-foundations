```python
import pandas as pd
import glob

# Get all parquet files in the current directory, excluding 'sample.parquet'
files = [f for f in glob.glob("*.parquet") if f != "sample.parquet"]

# Read and combine all parquet files
dfs = [pd.read_parquet(f) for f in files]
combined_df = pd.concat(dfs, ignore_index=True)

# Export combined dataframe to a new parquet file
combined_df.to_parquet("sample.parquet", index=False)

print(f"Combined {len(files)} parquet files into sample.parquet")
```
